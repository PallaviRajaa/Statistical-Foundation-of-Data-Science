```python
import pandas as pd
import numpy as np
data = {
    "Name": ["Pallavi", "Akshita", "Ritika", "Ishu"],
    "Age": [20, 21, 22, 23],
    "Marks": [88, 54, 63, 100],
    "Gender": ["Female", "Male", "Male", "Male"]
}
df = pd.DataFrame(data)
A = df[["Age", "Marks"]].to_numpy()

print("NumPy Array:")
print(A)
```

```python
print("Shape:", A.shape)
```

```python
print("Size:", A.size)
```

```python
print("Dimension:", A.ndim)
```

```python
B = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

print("Matrix B:")
print(B)
```

```python
print("Addition (A + B):")
print(A + B)
```

```python
print("Subtraction (A - B):")
print(A - B)
```

```python
print("Element-wise Multiplication (A * B):")
print(A * B)
```

```python
print("Matrix Product (A.T @ B):")
print(np.dot(A.T, B))
```

```python
print("Transpose of A:")
print(A.T)
```

```python
C = np.array([
    [2, 1],
    [5, 3]
])

print("Square Matrix C:")
print(C)

print("Inverse of C:")
print(np.linalg.inv(C))
```
