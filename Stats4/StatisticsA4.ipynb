**Imported libraries needed for data** **analysis and**
**visualization**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as pyplot
import seaborn as sns
```

**Identify duplicate cases using prof. Calculate average and standard deviation of age.**

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ST0151EN-SkillsNetwork/labs/teachingratings.csv"

ratings_df = pd.read_csv(url)

ratings_df.head()

# Identify duplicate cases using professor ID
print("Duplicate cases:")
print(ratings_df[ratings_df.duplicated(subset="prof", keep=False)].sort_values("prof"))

# Average and standard deviation using all observations
print("\nUsing all observations:")
print("Average age:", ratings_df["age"].mean())
print("Standard deviation:", ratings_df["age"].std())

# Keep only one observation for each professor
unique_professors = ratings_df.drop_duplicates(subset="prof")

print("\nNumber of unique professors:", len(unique_professors))

# Average and standard deviation using unique professors
print("\nUsing one observation per professor:")
print("Average age:", unique_professors["age"].mean())
print("Standard deviation:", unique_professors["age"].std())
```

**Question 2: Using a bar chart, demonstrate whether instructors teaching lower-division courses receive higher average evaluations.**

```python

# Calculate average evaluation by course division
average_eval = ratings_df.groupby("division")["eval"].mean()

print(average_eval)

# Create bar chart
plt.bar(average_eval.index, average_eval.values)

plt.xlabel("Course Division")
plt.ylabel("Average Teaching Evaluation")
plt.title("Average Evaluation by Course Division")

plt.show()
```

The bar chart compares the average teaching evaluation scores of instructors teaching different course divisions. Compare the bar for lower with the other division to determine whether lower-division instructors receive higher average evaluations.

**Question 3: Plot the relationship between age and teaching evaluation scores.**

```python

plt.scatter(ratings_df["age"], ratings_df["eval"])

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation Score")
plt.title("Age vs Teaching Evaluation")

plt.show()
```

The scatter plot shows the relationship between professors' ages and their teaching evaluation scores. Each point represents an observation.

**Question 4: Using gender-differentiated scatter plots, plot age against teaching evaluation scores.**

```python


# Separate data by gender
female = ratings_df[ratings_df["gender"] == "female"]
male = ratings_df[ratings_df["gender"] == "male"]

# Create scatter plot
plt.scatter(female["age"], female["eval"], label="Female")
plt.scatter(male["age"], male["eval"], label="Male")

plt.xlabel("Age")
plt.ylabel("Teaching Evaluation Score")
plt.title("Age vs Evaluation by Gender")

plt.legend()
plt.show()
```

The scatter plot uses different colors to represent male and female professors. This helps compare the relationship between age and evaluation scores across genders.

**Question 5: Create a scatterplot of age and evaluation scores, differentiated by gender and tenure.**

```python

# Question 5

# Separate data by gender and tenure
female_tenured = ratings_df[
    (ratings_df["gender"] == "female") &
    (ratings_df["tenure"] == "yes")
]

female_untenured = ratings_df[
    (ratings_df["gender"] == "female") &
    (ratings_df["tenure"] == "no")
]

male_tenured = ratings_df[
    (ratings_df["gender"] == "male") &
    (ratings_df["tenure"] == "yes")
]

male_untenured = ratings_df[
    (ratings_df["gender"] == "male") &
    (ratings_df["tenure"] == "no")
]

# Create scatter plot
pyplot.scatter(female_tenured["age"], female_tenured["eval"],
            label="Female Tenured")

pyplot.scatter(female_untenured["age"], female_untenured["eval"],
            label="Female Untenured")

pyplot.scatter(male_tenured["age"], male_tenured["eval"],
            label="Male Tenured")

pyplot.scatter(male_untenured["age"], male_untenured["eval"],
            label="Male Untenured")

pyplot.xlabel("Age")
pyplot.ylabel("Teaching Evaluation Score")
pyplot.title("Age vs Evaluation by Gender and Tenure")

pyplot.legend()
pyplot.show()
```

This scatter plot differentiates professors according to both gender and tenure status. It helps compare evaluation scores among female and male professors who are tenured or untenured.
