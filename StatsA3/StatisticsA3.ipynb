```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ST0151EN-SkillsNetwork/labs/teachingratings.csv"

ratings_df = pd.read_csv(url)

ratings_df.head()
```

**Calculate the percentage of visible minorities who are tenure professors.**

```python
tenure_minority = ratings_df.groupby("minority")["tenure"].count()

percentage = 100 * tenure_minority / tenure_minority.sum()

print(percentage)
```

13.82% of the professors in the dataset are visible minorities, while 86.18% are non-minorities. Therefore, visible minority professors make up a smaller proportion of the dataset.

**Does average age differ by tenure?**

```python
ratings_df.groupby("tenure")["age"].agg(["mean", "std"])
```

The average age is different between the two groups. Untenured professors have an average age of about 50.19 years, while tenured professors have an average age of about 47.85 years.

**Which graph works better for the age variable?**

```python
plt.hist(ratings_df["age"])

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Professor Age")

plt.show()
```

A histogram is more suitable for the age variable because age is numerical. It helps us see the distribution and frequency of professors across different age ranges.


**Difference between pyplot.bar and pyplot.barh**

```python
gender_count = ratings_df["gender"].value_counts()

plt.bar(gender_count.index, gender_count.values)
plt.xlabel("Gender")
plt.ylabel("Number of Professors")
plt.title("Gender Distribution using bar()")
plt.show()
```

```python

gender_count = ratings_df["gender"].value_counts()
plt.barh(gender_count.index, gender_count.values)
plt.ylabel("Gender")
plt.xlabel("Number of Professors")
plt.title("Gender Distribution using barh()")
plt.show()
```

A bar graph is suitable for gender because gender is a categorical variable. pyplot.bar displays the categories vertically, while pyplot.barh displays them horizontally.


**What is the median evaluation score for tenured professors?**

```python
median_eval = ratings_df[ratings_df["tenure"] == "yes"]["eval"].median()

print("Median evaluation score:", median_eval)
```

The median evaluation score for tenured professors is the value produced by the code above.
