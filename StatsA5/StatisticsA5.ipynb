**Q1. Probability of receiving an evaluation score greater than 4.5**

```python
from scipy.stats import norm

mean = 3.998
std = 0.554

z = (4.5 - mean) / std

probability = 1 - norm.cdf(z)

print("Z-score:", z)
print("Probability:", probability)
print("Percentage:", probability * 100)
```

There is approximately an 18.24% probability of receiving a teaching evaluation score greater than 4.5.


**Q2. Probability of receiving an evaluation score greater than 3.5 and less than 4.2**

```python
from scipy.stats import norm

mean = 3.998
std = 0.554

z1 = (3.5 - mean) / std
z2 = (4.2 - mean) / std

probability = norm.cdf(z2) - norm.cdf(z1)

print("Z-score for 3.5:", z1)
print("Z-score for 4.2:", z2)
print("Probability:", probability)
print("Percentage:", probability * 100)
```

There is approximately a 46% probability of receiving an evaluation score between 3.5 and 4.2.


**Q3. Two-tailed test — Basketball players**

```python
from scipy.stats import norm

mean = 12
std = 5.5
sample_mean = 10.7
n = 36

se = std / (n ** 0.5)

z = (sample_mean - mean) / se

p_value = 2 * norm.cdf(-abs(z))

print("Standard Error:", se)
print("Z-score:", z)
print("P-value:", p_value)
```

Since the p-value (approximately 0.156) is greater than 0.05, we fail to reject the null hypothesis. There is not enough statistical evidence to conclude that the average points scored by the regional players is different from the historic professional mean of 12 points per game.
